```python
import paths_config
from pathlib import Path
import glob
from datetime import datetime

synthea_files = sorted(glob.glob(str(paths_config.DATA_PROCESSED / "synthea_data_*.txt")), reverse=True)
mimic_files = sorted(glob.glob(str(paths_config.DATA_PROCESSED / "mimic_data_*.txt")), reverse=True)
if not synthea_files:
    raise FileNotFoundError("Nie znaleziono synthea_data_*.txt w data/processed/")
if not mimic_files:
    raise FileNotFoundError("Nie znaleziono mimic_data_*.txt w data/processed/")
synthea_file = synthea_files[0]
mimic_file = mimic_files[0]
print(f"📂 Synthea: {Path(synthea_file).name}")
print(f"📂 MIMIC: {Path(mimic_file).name}")

output_file = str(paths_config.DATA_PROCESSED / f'corpus_combined_{datetime.now().strftime("%Y%m%d_%H%M%S")}.txt')
total_lines = 0

with open(output_file, 'w') as outfile:
    for fname in [synthea_file, mimic_file]:
        file_lines = 0
        with open(fname, 'r') as infile:
            for line in infile:
                line = line.strip()
                if line:  # Pomijamy puste linie
                    outfile.write(line + '\n')
                    file_lines += 1
                    total_lines += 1
        print(f"   ✅ {Path(fname).name}: {file_lines:,} linii")

print("\n✅ Scalono Synthea i MIMIC!")
print(f"📊 Plik wyjściowy: {output_file}")
print(f"📊 Łączna liczba sekwencji: {total_lines:,}")
```
